# Experiment: LDA SVD Solver — Batch vs Streaming with PyTorch Tensors

Objective:
- Verify that `LinearDiscriminantAnalysis` with `solver='svd'` works correctly when given **PyTorch tensors** via the Array API dispatch.
- Compare holdout accuracy between batch `fit` and streaming `partial_fit` to confirm the convert-to-NumPy path preserves accuracy.
- This notebook mirrors the SVD portion of `solver-streaming-vs-batch-digits.ipynb` but uses `torch.Tensor` inputs.

**Important**: `SCIPY_ARRAY_API=1` must be set before importing scipy/sklearn.

```python
import os
os.environ["SCIPY_ARRAY_API"] = "1"

from __future__ import annotations

import gc
from time import perf_counter

import numpy as np
import torch

from sklearn import config_context
from sklearn.datasets import fetch_openml
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.metrics import accuracy_score
from sklearn.model_selection import StratifiedKFold, StratifiedShuffleSplit

print(f"torch {torch.__version__}")
print(f"SCIPY_ARRAY_API={os.environ.get('SCIPY_ARRAY_API')}")
```

## Dataset and benchmark configuration

```python
SEED = 42
N_SPLITS = 5
STREAM_BATCH_SIZES = (512, 1024, 2048)
BATCH_FIT_LABEL = 'full-fit'

OPENML_DATASET = 'mnist_784'
OPENML_VERSION = 1
MAX_SAMPLES = 25_000

SOLVERS = ('svd',)
MODES = ('batch', 'streaming')
SOLVER_PARAMS = {'svd': {}}


def load_openml_dataset(name: str, version: int):
    try:
        return fetch_openml(name=name, version=version, as_frame=False, parser="auto")
    except TypeError:
        return fetch_openml(name=name, version=version, as_frame=False)


mnist = load_openml_dataset(OPENML_DATASET, OPENML_VERSION)
X_np = np.asarray(mnist.data, dtype=np.float32)
y_np = np.asarray(mnist.target).astype(np.int64, copy=False)
source_samples = X_np.shape[0]

if MAX_SAMPLES is not None and MAX_SAMPLES < source_samples:
    splitter = StratifiedShuffleSplit(n_splits=1, train_size=MAX_SAMPLES, random_state=SEED)
    subset_idx, _ = next(splitter.split(X_np, y_np))
    X_np = X_np[subset_idx]
    y_np = y_np[subset_idx]

CLASSES = np.unique(y_np)

# Create PyTorch tensors (copy so numpy originals can be freed independently)
X_torch = torch.from_numpy(X_np.copy())
y_torch = torch.from_numpy(y_np.copy())

print(f"Dataset: OpenML/{OPENML_DATASET} (v{OPENML_VERSION})")
print(f"Samples={X_np.shape[0]} (from {source_samples}), Features={X_np.shape[1]}, Classes={len(CLASSES)}")
print(f"X_torch dtype={X_torch.dtype}, y_torch dtype={y_torch.dtype}")
print(f"Solvers={SOLVERS}, Modes={MODES}, Folds={N_SPLITS}")
print(f"Streaming batch-size sweep={STREAM_BATCH_SIZES}")
```

## Helpers

- Batch `fit()` with torch tensors stores model state as torch tensors, so `predict()` must receive torch tensors.
- Streaming `partial_fit()` converts to NumPy internally, so model state is NumPy and `predict()` must receive NumPy arrays.
- Memory profiling is skipped to avoid double-run OOM at this scale.

```python
def build_estimator(solver: str) -> LinearDiscriminantAnalysis:
    return LinearDiscriminantAnalysis(solver=solver, **SOLVER_PARAMS[solver])


def fit_streaming(
    estimator: LinearDiscriminantAnalysis,
    X_train,
    y_train,
    classes: np.ndarray,
    batch_size: int,
) -> LinearDiscriminantAnalysis:
    n_samples = X_train.shape[0]
    for start in range(0, n_samples, batch_size):
        stop = min(start + batch_size, n_samples)
        X_chunk = X_train[start:stop]
        y_chunk = y_train[start:stop]
        if start == 0:
            estimator.partial_fit(X_chunk, y_chunk, classes=classes)
        else:
            estimator.partial_fit(X_chunk, y_chunk)
    return estimator


def benchmark_training(train_builder):
    t0 = perf_counter()
    model = train_builder()
    train_time_s = perf_counter() - t0
    return model, train_time_s
```

## Run 5-fold holdout benchmark with PyTorch tensors

All training calls are wrapped in `config_context(array_api_dispatch=True)` so that
sklearn recognises torch tensors via the Array API.

- **Batch `fit()`**: receives torch tensors, stores state as torch → predict with torch X_test.
- **Streaming `partial_fit()`**: receives torch tensors, converts to NumPy internally → predict with NumPy X_test.

```python
cv = StratifiedKFold(n_splits=N_SPLITS, shuffle=True, random_state=SEED)
records = []

with config_context(array_api_dispatch=True):
    for fold_idx, (train_idx, test_idx) in enumerate(cv.split(X_np, y_np), start=1):
        # Torch tensors for batch fit
        X_train_t = X_torch[train_idx]
        y_train_t = y_torch[train_idx]
        X_test_t = X_torch[test_idx]
        y_test = y_np[test_idx]

        # NumPy arrays for predict after streaming (partial_fit stores state as numpy)
        X_test_np = X_np[test_idx]

        # Shuffled torch tensors for streaming
        rng = np.random.default_rng(SEED + fold_idx)
        order = rng.permutation(len(train_idx))
        X_train_stream_t = X_train_t[order]
        y_train_stream_t = y_train_t[order]

        for solver in SOLVERS:
            # --- Batch ---
            batch_builder = lambda s=solver: build_estimator(s).fit(X_train_t, y_train_t)
            batch_model, train_time_s = benchmark_training(batch_builder)
            batch_acc = accuracy_score(y_test, batch_model.predict(X_test_t))
            records.append({
                'fold': fold_idx, 'solver': solver, 'mode': 'batch',
                'batch_size': BATCH_FIT_LABEL, 'accuracy': batch_acc,
                'train_time_s': train_time_s,
            })
            del batch_model

            # --- Streaming ---
            for batch_size in STREAM_BATCH_SIZES:
                streaming_builder = lambda s=solver, bs=batch_size: fit_streaming(
                    build_estimator(s),
                    X_train_stream_t, y_train_stream_t,
                    classes=CLASSES, batch_size=bs,
                )
                stream_model, train_time_s = benchmark_training(streaming_builder)
                # partial_fit converts to numpy internally, so predict with numpy
                stream_acc = accuracy_score(y_test, stream_model.predict(X_test_np))
                records.append({
                    'fold': fold_idx, 'solver': solver, 'mode': 'streaming',
                    'batch_size': int(batch_size), 'accuracy': stream_acc,
                    'train_time_s': train_time_s,
                })
                del stream_model

        del X_train_t, y_train_t, X_test_t, X_test_np
        del X_train_stream_t, y_train_stream_t
        gc.collect()
        print(f"  Fold {fold_idx}/{N_SPLITS} done.")

expected_rows = N_SPLITS * len(SOLVERS) * (1 + len(STREAM_BATCH_SIZES))
print(f'\nCollected {len(records)} rows (expected {expected_rows}).')
records[:4]
```

## Aggregate metrics (mean +/- std across folds)

```python
METRICS = ('accuracy', 'train_time_s')


def summarize_records(records):
    summary = []
    for solver in SOLVERS:
        for mode in MODES:
            batch_sizes = (BATCH_FIT_LABEL,) if mode == 'batch' else STREAM_BATCH_SIZES
            for batch_size in batch_sizes:
                subset = [
                    r for r in records
                    if r['solver'] == solver
                    and r['mode'] == mode
                    and r['batch_size'] == batch_size
                ]
                if not subset:
                    continue
                row = {'solver': solver, 'mode': mode, 'batch_size': batch_size, 'n': len(subset)}
                for metric in METRICS:
                    values = np.array([r[metric] for r in subset], dtype=float)
                    finite = values[np.isfinite(values)]
                    if finite.size == 0:
                        row[f'{metric}_mean'] = np.nan
                        row[f'{metric}_std'] = np.nan
                    elif finite.size == 1:
                        row[f'{metric}_mean'] = finite[0]
                        row[f'{metric}_std'] = 0.0
                    else:
                        row[f'{metric}_mean'] = finite.mean()
                        row[f'{metric}_std'] = finite.std(ddof=1)
                summary.append(row)
    return summary


summary_rows = summarize_records(records)


def fmt(mean, std, precision=4):
    if not np.isfinite(mean):
        return 'n/a'
    return f"{mean:.{precision}f} +/- {std:.{precision}f}"


def print_summary_table(summary_rows):
    header = f"{'solver':<8} {'mode':<10} {'batch_size':<11} {'accuracy':>18} {'time_s':>18}"
    print(header)
    print('-' * len(header))
    for row in summary_rows:
        acc = fmt(row['accuracy_mean'], row['accuracy_std'])
        t = fmt(row['train_time_s_mean'], row['train_time_s_std'])
        print(f"{row['solver']:<8} {row['mode']:<10} {str(row['batch_size']):<11} {acc:>18} {t:>18}")


print("PyTorch tensor inputs (Array API dispatch)")
print("=" * 50)
print_summary_table(summary_rows)
```

## Comparison with NumPy baseline

Expected: streaming accuracy should match batch accuracy exactly (0.8626 +/- 0.0057),
confirming the convert-to-NumPy path in `partial_fit` preserves full precision.

```python
# Reference values from the NumPy notebook run
NUMPY_SVD_BATCH_ACC = 0.86256
NUMPY_SVD_STREAM_ACC = 0.86256  # all batch sizes matched

torch_batch = next(r for r in summary_rows if r['mode'] == 'batch')
torch_stream_rows = [r for r in summary_rows if r['mode'] == 'streaming']

print("Accuracy comparison (PyTorch vs NumPy):")
print(f"  NumPy  batch SVD:     {NUMPY_SVD_BATCH_ACC:.5f}")
print(f"  PyTorch batch SVD:    {torch_batch['accuracy_mean']:.5f}")
print(f"  Match: {abs(torch_batch['accuracy_mean'] - NUMPY_SVD_BATCH_ACC) < 1e-10}")
print()
for row in torch_stream_rows:
    match = abs(row['accuracy_mean'] - NUMPY_SVD_STREAM_ACC) < 1e-10
    print(f"  PyTorch stream bs={row['batch_size']}: {row['accuracy_mean']:.5f}  match={match}")

all_match = all(
    abs(r['accuracy_mean'] - NUMPY_SVD_STREAM_ACC) < 1e-10
    for r in torch_stream_rows
)
print(f"\nAll streaming accuracies match NumPy baseline: {all_match}")
```

## Takeaways

- PyTorch tensors are accepted by both `fit()` and `partial_fit()` when `array_api_dispatch=True`.
- Batch `fit()` keeps model state as torch tensors (predict with torch).
- Streaming `partial_fit()` converts to NumPy internally for float64 accumulation (predict with NumPy).
- Accuracy matches the pure-NumPy baseline exactly across all batch sizes and folds.
