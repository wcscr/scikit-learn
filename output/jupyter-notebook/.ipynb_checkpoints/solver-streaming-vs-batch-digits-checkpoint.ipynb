# Experiment: LDA Solvers in Batch vs Streaming Mode (Digits Dataset)

Objective:
- Compare `LinearDiscriminantAnalysis` solvers (`svd`, `lsqr`, `eigen`) in full-batch (`fit`) and streaming (`partial_fit`) training.
- Measure training speed, peak Python memory, and holdout accuracy with 5-fold stratified cross-validation.
- Use a real scikit-learn dataset larger than Iris: `load_digits` (1,797 samples, 64 features, 10 classes).


```python
from __future__ import annotations

import tracemalloc
from time import perf_counter

import matplotlib.pyplot as plt
import numpy as np

from sklearn.datasets import load_digits
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.metrics import accuracy_score
from sklearn.model_selection import StratifiedKFold
```

## Dataset and benchmark configuration

This setup keeps the training protocol fixed while changing only the LDA solver and training mode.


```python
SEED = 42
N_SPLITS = 5
STREAM_BATCH_SIZE = 128

SOLVERS = ('svd', 'lsqr', 'eigen')
MODES = ('batch', 'streaming')
# Mild shrinkage improves numerical stability for covariance-based solvers.
SOLVER_PARAMS = {
    'svd': {},
    'lsqr': {'shrinkage': 1e-3},
    'eigen': {'shrinkage': 1e-3},
}

digits = load_digits()
X = digits.data.astype(np.float64, copy=False)
y = digits.target
CLASSES = np.unique(y)

print(f'Dataset: digits')
print(f'Samples={X.shape[0]}, Features={X.shape[1]}, Classes={len(CLASSES)}')
print(f'Solvers={SOLVERS}, Modes={MODES}, Folds={N_SPLITS}')
```

## Helpers

- `fit` mode trains once on the whole training fold.
- `partial_fit` mode streams mini-batches through the same fold.
- `tracemalloc` captures peak Python allocation during training.


```python
def build_estimator(solver: str) -> LinearDiscriminantAnalysis:
    return LinearDiscriminantAnalysis(solver=solver, **SOLVER_PARAMS[solver])


def fit_streaming(
    estimator: LinearDiscriminantAnalysis,
    X_train: np.ndarray,
    y_train: np.ndarray,
    classes: np.ndarray,
    batch_size: int,
) -> LinearDiscriminantAnalysis:
    for start in range(0, X_train.shape[0], batch_size):
        stop = min(start + batch_size, X_train.shape[0])
        X_chunk = X_train[start:stop]
        y_chunk = y_train[start:stop]
        if start == 0:
            estimator.partial_fit(X_chunk, y_chunk, classes=classes)
        else:
            estimator.partial_fit(X_chunk, y_chunk)
    return estimator


def timed_training(train_fn):
    tracemalloc.start()
    t0 = perf_counter()
    estimator = train_fn()
    elapsed = perf_counter() - t0
    _, peak_bytes = tracemalloc.get_traced_memory()
    tracemalloc.stop()
    return estimator, elapsed, peak_bytes / (1024 ** 2)
```

## Run 5-fold holdout benchmark


```python
cv = StratifiedKFold(n_splits=N_SPLITS, shuffle=True, random_state=SEED)
records = []

for fold_idx, (train_idx, test_idx) in enumerate(cv.split(X, y), start=1):
    X_train, y_train = X[train_idx], y[train_idx]
    X_test, y_test = X[test_idx], y[test_idx]

    rng = np.random.default_rng(SEED + fold_idx)
    order = rng.permutation(X_train.shape[0])
    X_train_stream = X_train[order]
    y_train_stream = y_train[order]

    for solver in SOLVERS:
        estimator = build_estimator(solver)
        batch_model, train_time_s, peak_mem_mb = timed_training(
            lambda est=estimator: est.fit(X_train, y_train)
        )
        batch_acc = accuracy_score(y_test, batch_model.predict(X_test))
        records.append(
            {
                'fold': fold_idx,
                'solver': solver,
                'mode': 'batch',
                'accuracy': batch_acc,
                'train_time_s': train_time_s,
                'peak_mem_mb': peak_mem_mb,
            }
        )

        estimator = build_estimator(solver)
        stream_model, train_time_s, peak_mem_mb = timed_training(
            lambda est=estimator: fit_streaming(
                est,
                X_train_stream,
                y_train_stream,
                classes=CLASSES,
                batch_size=STREAM_BATCH_SIZE,
            )
        )
        stream_acc = accuracy_score(y_test, stream_model.predict(X_test))
        records.append(
            {
                'fold': fold_idx,
                'solver': solver,
                'mode': 'streaming',
                'accuracy': stream_acc,
                'train_time_s': train_time_s,
                'peak_mem_mb': peak_mem_mb,
            }
        )

print(f'Collected {len(records)} rows ({N_SPLITS} folds x {len(SOLVERS)} solvers x 2 modes).')
records[:4]
```

## Aggregate metrics (mean +/- std across folds)


```python
METRICS = ('accuracy', 'train_time_s', 'peak_mem_mb')


def summarize_records(records):
    summary = []
    for solver in SOLVERS:
        for mode in MODES:
            subset = [r for r in records if r['solver'] == solver and r['mode'] == mode]
            if not subset:
                continue
            row = {'solver': solver, 'mode': mode, 'n': len(subset)}
            for metric in METRICS:
                values = np.array([r[metric] for r in subset], dtype=float)
                row[f'{metric}_mean'] = values.mean()
                row[f'{metric}_std'] = values.std(ddof=1) if len(values) > 1 else 0.0
            summary.append(row)
    return summary


summary_rows = summarize_records(records)
summary_rows
```

```python
def print_summary_table(summary_rows):
    header = (
        f"{'solver':<8} {'mode':<10} {'accuracy':>18} {'train_time_s':>20} {'peak_mem_mb':>20}"
    )
    print(header)
    print('-' * len(header))
    for row in summary_rows:
        acc = f"{row['accuracy_mean']:.4f} +/- {row['accuracy_std']:.4f}"
        t = f"{row['train_time_s_mean']:.4f} +/- {row['train_time_s_std']:.4f}"
        m = f"{row['peak_mem_mb_mean']:.2f} +/- {row['peak_mem_mb_std']:.2f}"
        print(f"{row['solver']:<8} {row['mode']:<10} {acc:>18} {t:>20} {m:>20}")


print_summary_table(summary_rows)
```

## Visual comparison


```python
summary_lookup = {(r['solver'], r['mode']): r for r in summary_rows}


def plot_metric(metric: str, ylabel: str):
    x = np.arange(len(SOLVERS))
    width = 0.36

    fig, ax = plt.subplots(figsize=(8.5, 4.0))
    for i, mode in enumerate(MODES):
        means = [summary_lookup[(solver, mode)][f'{metric}_mean'] for solver in SOLVERS]
        stds = [summary_lookup[(solver, mode)][f'{metric}_std'] for solver in SOLVERS]
        centers = x + (i - 0.5) * width
        ax.bar(centers, means, width=width, yerr=stds, capsize=4, label=mode)

    ax.set_title(f'LDA {metric} by solver and mode')
    ax.set_xlabel('Solver')
    ax.set_ylabel(ylabel)
    ax.set_xticks(x)
    ax.set_xticklabels(SOLVERS)
    ax.legend(title='Training mode')
    ax.grid(alpha=0.25, axis='y')
    plt.tight_layout()
    plt.show()


plot_metric('train_time_s', 'Training time (seconds)')
plot_metric('peak_mem_mb', 'Peak Python memory (MB)')
plot_metric('accuracy', 'Holdout accuracy')
```

## Quick takeaways from this run


```python
def best(summary_rows, metric: str, higher_is_better: bool):
    key = f'{metric}_mean'
    fn = max if higher_is_better else min
    return fn(summary_rows, key=lambda row: row[key])


best_acc = best(summary_rows, 'accuracy', higher_is_better=True)
fastest = best(summary_rows, 'train_time_s', higher_is_better=False)
lowest_mem = best(summary_rows, 'peak_mem_mb', higher_is_better=False)

print('Best accuracy:', best_acc['solver'], best_acc['mode'], f"({best_acc['accuracy_mean']:.4f})")
print('Fastest training:', fastest['solver'], fastest['mode'], f"({fastest['train_time_s_mean']:.4f} s)")
print('Lowest peak memory:', lowest_mem['solver'], lowest_mem['mode'], f"({lowest_mem['peak_mem_mb_mean']:.2f} MB)")
```
